# SEC filings analysis

Retrieve the latest income statement, balance sheet, and cash-flow data available for a company ticker through Yahoo Finance.

```python
import yfinance as yf
```

```python
def get_financials(ticker):
    """Print the latest available financial statements for a ticker."""
    stock = yf.Ticker(ticker)

    print("\n=== Income Statement ===\n")
    print(stock.financials.head())

    print("\n=== Balance Sheet ===\n")
    print(stock.balance_sheet.head())

    print("\n=== Cash Flow ===\n")
    print(stock.cashflow.head())
```

```python
get_financials("MU")
```
