# Stock price and analyst ratings

Look up a current price and the latest analyst recommendation history for a ticker.

```python
import yfinance as yf

def get_price(ticker):
    """Print the current price reported for a ticker."""
    price = yf.Ticker(ticker).info.get("currentPrice")
    print(f"{ticker} current price: {price}")
```

```python
get_price("MU")
```

```python
def get_analyst_ratings(ticker):
    """Print the ten most recent analyst recommendations."""
    recommendations = yf.Ticker(ticker).recommendations

    if recommendations is None or recommendations.empty:
        print(f"No analyst recommendations found for {ticker}")
        return

    print(f"\nAnalyst recommendations for {ticker}:\n")
    print(recommendations.tail(10))
```

```python
get_analyst_ratings("MU")
```
